```python
# importar bibliotecas
%pip install openpyxl
import json
import pandas as pd
```

```python

# 1. Cargar clientes (revisa si el archivo realmente se llama '.csv.csv')
clientes = pd.read_csv('./dataset/ventas.csv')

# 2. Cargar opiniones desde JSON
with open('./dataset/opiniones.json', 'r', encoding='utf-8') as f:
    datos_json = json.load(f)
opiniones = pd.DataFrame(datos_json)

# 3. Cargar productos desde Excel (corregido para usar solo una variable)
productos = pd.read_excel('./dataset/productos.xlsx')
```

```python
# 1. Unir 'ventas' con 'productos' usando 'producto_id'
# Ambos archivos comparten exactamente la columna 'producto_id'
df_temporal = pd.merge(ventas, productos, on="producto_id", how="left")

# 2. Unir el resultado anterior con 'opiniones' usando también 'producto_id'
# Esto asociará las opiniones de la red social a cada producto vendido
df_final = pd.merge(df_temporal, opiniones, on="producto_id", how="left")
```
